# Week 1 Assignment: Titanic Survival Prediction
## Data Cleaning & Preprocessing

**Objective:** Clean and preprocess the Titanic dataset by inspecting the data, handling missing values, encoding categorical variables, visualizing age distribution, and exporting the cleaned dataset.

**Dataset:** Titanic-Dataset.csv

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder

pd.set_option('display.max_columns', None)
```

## 1. Load the Dataset

```python

titanic = pd.read_csv('Titanic-Dataset.csv')

print(f'Dataset shape: {titanic.shape}')
titanic.head()
```

## 2. Basic Dataset Information

```python

titanic.info()
```

```python

titanic.describe(include='all').T
```

## 3. Identify Missing Data

```python
missing_data = pd.DataFrame({
    'Missing Count': titanic.isnull().sum(),
    'Missing Percentage': (titanic.isnull().mean() * 100).round(2)
}).sort_values('Missing Count', ascending=False)

missing_data
```

## 4. Handle Missing Values

- **Age:** Fill missing values with the median age because it is less affected by extreme values.
- **Embarked:** Fill the two missing values with the most frequent port.
- **Cabin:** Drop the column because most of its values are missing, making reliable imputation impractical for this preprocessing task.

```python

titanic['Age'] = titanic['Age'].fillna(titanic['Age'].median())

titanic['Embarked'] = titanic['Embarked'].fillna(titanic['Embarked'].mode()[0])

titanic = titanic.drop(columns=['Cabin'])

print('Missing values after handling:')
print(titanic.isnull().sum())
```

## 5. Encode Categorical Variables

```python

sex_encoder = LabelEncoder()
titanic['Sex'] = sex_encoder.fit_transform(titanic['Sex'])

sex_mapping = dict(zip(
    sex_encoder.classes_,
    sex_encoder.transform(sex_encoder.classes_)
))
print('Sex encoding:', sex_mapping)

titanic = pd.get_dummies(
    titanic,
    columns=['Embarked'],
    drop_first=True,
    dtype=int
)

titanic.head()
```

## 6. Visualize Age Distribution

```python
plt.figure(figsize=(9, 5))
sns.histplot(data=titanic, x='Age', bins=30, kde=True)
plt.title('Age Distribution of Titanic Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')
plt.tight_layout()
plt.show()
```

## 7. Validate the Cleaned Dataset

```python
print('Final dataset shape:', titanic.shape)
print('Total missing values:', int(titanic.isnull().sum().sum()))
print('\nFinal columns:')
print(titanic.columns.tolist())

assert titanic.isnull().sum().sum() == 0
print('\nValidation passed: the cleaned dataset contains no missing values.')
```

## 8. Export the Cleaned Dataset

```python
output_file = 'titanic_cleaned.csv'
titanic.to_csv(output_file, index=False)

print(f'Cleaned dataset exported successfully as: {output_file}')
```

## Conclusion

The Titanic dataset was successfully cleaned and preprocessed. Missing values were handled appropriately, categorical variables were encoded, the age distribution was visualized, and the final cleaned dataset was exported as `titanic_cleaned.csv`.

This preprocessing step prepares the dataset for further machine learning tasks such as feature selection, model training, and Titanic survival prediction.
